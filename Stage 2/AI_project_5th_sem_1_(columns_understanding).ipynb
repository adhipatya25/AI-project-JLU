```python
numerical_cols = df.select_dtypes(include=np.number).columns
categorical_cols = df.select_dtypes(include="object").columns
```

```python
print(numerical_cols)
print(categorical_cols)
```

```python
zero_counts = {}
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        zero_counts[col] = (df[col] == 0).sum()
    elif df[col].dtype == 'object':
        zero_counts[col] = ((df[col] == '0') | (df[col] == 0)).sum()

zero_df = pd.DataFrame(zero_counts.items(), columns=['Column', 'Zero_Count'])
zero_df['Zero_Percentage'] = (zero_df['Zero_Count'] / len(df)) * 100

print("Columns with zero values and their percentages:")
print(zero_df[zero_df['Zero_Count'] > 0])
```
