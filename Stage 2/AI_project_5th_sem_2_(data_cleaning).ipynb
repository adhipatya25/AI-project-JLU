```python
import pandas as pd

# Load Dataset

df = pd.read_csv("/content/drive/MyDrive/AI-Project-5th-Sem/merged_dataset_with_target_final.csv")

print("="*60)
print("ORIGINAL DATASET")
print("="*60)

print("Shape :", df.shape)
print()

# Store original information
original_rows = df.shape[0]
original_cols = df.shape[1]

# Columns to Remove

columns_to_delete = [

    "player_id",
    "first_name",
    "last_name",
    "name_x",
    "last_season_x",
    "current_club_id",
    "player_code",
    "country_of_birth",
    "city_of_birth",
    "agent_name",
    "image_url",
    "current_national_team_id",
    "url_x",
    "current_value",
    "highest_value",
    "average_value",
    "min_value",
    "club_id",
    "club_code",
    "name_y",
    "domestic_competition_id",
    "total_market_value",
    "coach_name",
    "last_season_y",
    "filename",
    "url_y",
    "competition_id",
    "country_id",
    "url"

]

# Delete only existing columns
existing_columns = [col for col in columns_to_delete if col in df.columns]

df.drop(columns=existing_columns, inplace=True)

# Reset Index

df.reset_index(drop=True, inplace=True)

# Final Dataset Information

print("="*60)
print("DATA CLEANING SUMMARY")
print("="*60)

print(f"Original Rows                : {original_rows}")
print(f"Original Columns             : {original_cols}")

print()

print(f"Columns Removed              : {len(existing_columns)}")
print()

print(f"Final Rows                   : {df.shape[0]}")
print(f"Final Columns                : {df.shape[1]}")

print("="*60)

# Remaining Columns

print("\nRemaining Columns:\n")

for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")

# Save Cleaned Dataset

output_file = "cleaned_football_dataset_final.csv"

df.to_csv(output_file, index=False)

print()
print("="*60)
print(f"Cleaned dataset saved successfully as:")
print(output_file)
print("="*60)
```
