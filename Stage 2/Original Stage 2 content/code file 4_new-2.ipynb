```python
# T-Test coding
# Is there a significant difference in Future Potential Scores between Attack and Defender players?

import pandas as pd
from scipy import stats

# Load dataset
df = pd.read_csv("/content/drive/MyDrive/AI-Project-5th-Sem/cleaned_football_dataset_final.csv")

# Select the two groups
attack = df[df["position"] == "Attack"]["Player_Future_Potential_Score"].dropna()

defender = df[df["position"] == "Defender"]["Player_Future_Potential_Score"].dropna()

# Significance level
alpha = 0.05

# Independent two-sample t-test
t_stat, p_value = stats.ttest_ind(attack, defender, equal_var=True)

print("Attack mean:", attack.mean())
print("Defender mean:", defender.mean())
print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < alpha:
   print("We reject the null hypothesis")
else:
    print("We fail to reject the null hypothesis")
```

The p-value is less than 0.05, so we reject the null hypothesis. There is a statistically significant difference between the average Future Potential Scores of Attack and Defender players.

```python
# ANOVA Coding
# Is there a significant difference in Future Potential Scores among Attack, Midfield, and Defender players?

import pandas as pd
from scipy import stats

# Load dataset
df = pd.read_csv('/content/drive/MyDrive/AI-Project-5th-Sem/cleaned_football_dataset_final.csv')

# Select the three groups
attack = df[df["position"] == "Attack"]["Player_Future_Potential_Score"].dropna()
midfield = df[df["position"] == "Midfield"]["Player_Future_Potential_Score"].dropna()
defender = df[df["position"] == "Defender"]["Player_Future_Potential_Score"].dropna()

# Significance level
alpha = 0.05

# One-Way ANOVA
f_stat, p_value = stats.f_oneway(attack, midfield, defender)

print("Attack mean:", attack.mean())
print("Midfield mean:", midfield.mean())
print("Defender mean:", defender.mean())
print("F-statistic:", f_stat)
print("p-value:", p_value)

# Decision
if p_value < alpha:
    print("We reject the null hypothesis")
else:
    print("We fail to reject the null hypothesis")
```

The p-value is less than 0.05, so we reject the null hypothesis. The mean Future Potential Scores are not the same across Attack, Midfield, and Defender players.

```python
# Chi-square Test Coding
# Is player position associated with Future Potential Rating?

import pandas as pd
from scipy import stats

# Load dataset
df = pd.read_csv('/content/drive/MyDrive/AI-Project-5th-Sem/cleaned_football_dataset_final.csv')

# Convert the text placeholder "Missing" into a real missing value
df["position"] = df["position"].replace("Missing", pd.NA)

# Remove rows with missing position before creating the contingency table
df = df.dropna(subset=["position", "Player_Future_Potential_Rating"])

# Create contingency table
table = pd.crosstab(
    df["position"],
    df["Player_Future_Potential_Rating"]
)

print("Contingency Table:")
print(table)

# Chi-square test
chi2, p_value, dof, expected = stats.chi2_contingency(table)

# Significance level
alpha = 0.05

print("\nChi-square statistic:", chi2)
print("p-value:", p_value)
print("Degrees of freedom:", dof)

# Decision
if p_value < alpha:
    print("We reject the null hypothesis")
else:
    print("We fail to reject the null hypothesis")
```

The p-value is less than 0.05, so we reject the null hypothesis. There is a statistically significant association between player position and Future Potential Rating.
